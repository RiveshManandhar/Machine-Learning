```python
import pandas as pd
import numpy
import matplotlib.pyplot as plt
from sklearn import linear_model 
```

```python
df=pd.read_excel('./data/homeprices.xlsx')
df
```

```python
%matplotlib inline
plt.xlabel('Area (sq ft)')
plt.xlabel('Price (USD)')
plt.scatter(df['area'],df['price'],color='red',marker='+')
```

```python
reg=linear_model.LinearRegression()
reg.fit(df[['area']],df.price)
```

```python
reg.predict(numpy.array([3300]).reshape(-1, 1))
```

```python
df2=pd.read_excel('./data/homepricesmultivariables.xlsx')
```

```python
#Handle missing data points
import math

median=math.floor(df2['bedrooms'].median())

df2['bedrooms']=df2['bedrooms'].fillna(median)

df2
```

```python
print(df2.columns)
reg=linear_model.LinearRegression()
reg.fit(df2[['area', 'bedrooms', 'age']], df2['price'])
```

```python
print(reg.intercept_)
print(reg.coef_)

```

```python
reg.predict([[3000, 3, 40]])
```

```python
df3=pd.read_excel('./data/salarymultivariables.xlsx')
df3['experience']=df3['experience'].fillna('zero')
```

```python
from word2number import w2n
df3['experience']=df3['experience'].apply(w2n.word_to_num)
```

```python
df3['test_score']=df3['test_score'].fillna(df3['test_score'].median())
```

```python
df3
```

```python
reg3=linear_model.LinearRegression()
reg3.fit(df3[['experience', 'test_score', 'interview_score']], df3['salary'])
```

```python
reg3.predict([[2, 9, 6]])
```

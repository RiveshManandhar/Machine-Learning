DistPlot (Numerical - Categorical)

HeatMap (Categorical - Categorical)

```python
#how many people survived and died in each pclass?

pd.crosstab(df['2urvived'],df['Pclass'])
```

```python
sns.heatmap(pd.crosstab(df['2urvived'],df['Pclass']))
```
